```python
from pathlib import Path

import ollama
```

```python
# model = "gemma4:e4b"
model = "gemma4:12b"
image_path = Path(r"C:\Users\oloon\Dropbox\images\photography\owl_girl.jpg")

response = ollama.chat(
    model=model,
    messages=[
        {
            "role": "user",
            "content": "Describe this image in one concise paragraph.",
            "images": [image_path],
        }
    ],
    options={"temperature": 0},
)

print(response["message"]["content"])
```
